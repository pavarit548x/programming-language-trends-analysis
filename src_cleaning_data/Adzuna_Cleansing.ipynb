```python
import re
from collections import Counter

lang_patterns = {k.lower(): k for k in languages_data.keys()}
COMBINED_LANG = "C++"

def normalize_description(text):
    if not isinstance(text, str):
        return ""
    t = " " + text.lower() + " "
    for wrong, correct in description_typo_map.items():
        t = t.replace(wrong, correct)
    return t

def find_languages_in_desc(text):
    if not isinstance(text, str):
        return set()
    
    text_lower = " " + text.lower() + " "
    found = set()
    
    # ────────────────────────────────────────
    # Special symbol languages — handle first
    # ────────────────────────────────────────
    has_csharp = bool(re.search(r'(?<!\w)c#(?!\w)', text_lower))
    if has_csharp:
        found.add('C#')
    
    has_cpp        = bool(re.search(r'(?<!\w)c\+\+(?!\w)', text_lower))
    has_isolated_c = bool(re.search(r'(?<!\w)c(?![\+#\w])', text_lower))
    if has_cpp or has_isolated_c:
        found.add(COMBINED_LANG)
    
    has_dotnet = bool(re.search(r'(?<!\w)\.net(?!\w)', text_lower))
    if has_dotnet:
        found.add('C#')

    # ────────────────────────────────────────
    # All other languages — general loop
    # ────────────────────────────────────────
    SYMBOL_LANGS = {'c', 'c++', 'c/c++', 'c#', '.net'}
    
    for lang_lower, original_name in lang_patterns.items():
        if lang_lower in SYMBOL_LANGS:
            continue
        if re.search(r'(?<!\w)' + re.escape(lang_lower) + r'(?!\w)', text_lower):
            found.add(original_name)
    
    return found

# ────────────────────────────────────────────────────────
# Main Loop — collect id, company, title, year, languages
# ────────────────────────────────────────────────────────
lang_counter = Counter()
results = []                    # ← store per-row data here

for _, row in df_Adzuna_clean.iterrows():   # ← iterate rows not just description
    
    # ── Extract fields ──────────────────────────────────
    job_id  = row.get('id', None)
    company = row.get('company', None)
    title   = row.get('title', None)
    year    = pd.to_datetime(row.get('created', None)).year  # extract year from datetime

    # ── Detect languages ────────────────────────────────
    normalized = normalize_description(row.get('description', ''))
    detected   = find_languages_in_desc(normalized)
    
    # ── Update counter ──────────────────────────────────
    lang_counter.update(detected)
    
    # ── Store result ─────────────────────────────────────
    results.append({
        'id'       : job_id,
        'company'  : company,
        'title'    : title,
        'year'     : year,
        'languages': list(detected)     # set → list for DataFrame
    })

# ────────────────────────────────────────
# Build Result DataFrame
# ────────────────────────────────────────
df_results = pd.DataFrame(results)

print("=== Job Language Detection Results ===")
display(df_results.head(10))

# ────────────────────────────────────────
# Language Frequency
# ────────────────────────────────────────
print("\nLanguage frequency in job descriptions:")
for lang, count in lang_counter.most_common():
    print(f"{lang} : {count}")
```
