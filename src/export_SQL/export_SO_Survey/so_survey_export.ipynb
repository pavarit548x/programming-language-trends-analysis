```python
import pandas as pd
import numpy as np
import json
from sqlalchemy import create_engine
import dotenv
dotenv.load_dotenv()

# 1. เชื่อมต่อกับฐานข้อมูล
# เปลี่ยน user, password, host, port, db_name ให้ตรงกับของคุณ
engine = create_engine(f'postgresql://postgres:{dotenv.dotenv_values().get("PASSWORD")}@localhost:5432/{dotenv.dotenv_values().get("DB_NAME")}')
```

```python
df_prolang = pd.read_sql('select * from prolang', engine)
for i in range(2020,2026):
    with open(f'SO_survey_{i}_desire.json', 'r', encoding='utf-8') as f:
        data = json.load(f)
    df = pd.DataFrame(data)
    # 1. Split ภาษาออกเป็น List
    df['Language'] = df['Language'].str.split('/')

    # 2. หาร count ด้วยจำนวนสมาชิกใน List
    # ใช้ .str.len() เพื่อดูว่ามีกี่ภาษาในแถวนั้น
    df['count'] = df['count'] / df['Language'].str.len()
    df['count'] = df['count'].astype('int64')

    # 3. Explode ให้กระจายเป็นแถวละภาษา
    df_clean = df.explode('Language')
    def mapping(val):
        if val == 'VISUAL BASIC (.NET)':
            return 'VB.NET'
        elif val == 'SHELL (ALL SHELLS)':
            return 'SHELL'
        return val
    df_clean['Language'] = df_clean['Language'].apply(mapping)
    df_survey = pd.merge(df_clean, df_prolang, left_on='Language', right_on='prolangname', how='left')
    df_survey.drop_duplicates(keep='first', inplace=True)
    df_survey = df_survey[['prolangid','count','year_of_survey']].rename(columns={'count':'total','year_of_survey':'year'})
    df_survey['year'] = df_survey['year'].astype(str)
    df_survey.sort_values(by='prolangid',inplace=True)
    print(df_survey.info())
    display(df_survey)
    df_survey.to_sql('so_survey', engine, if_exists='append', index=False)
```
